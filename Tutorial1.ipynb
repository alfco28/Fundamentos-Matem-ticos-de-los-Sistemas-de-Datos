## El programa `CMD.exe Prompt`

Este programa se encuentra en el directorio en el que hemos instalado nuestro `Anaconda`. Para ejecutarlo basta hacer doble click sobre programa y nos aparecerá una ventana del siguiente tipo
<img src="CMDexe.jpg" alt="cmd" style="width: 100%;"/>

Algunos de los comandos que tenemos son los siguientes:
* `dir` (o `ls`): muestra las carpetas (aquellos que aparecen junto a `<DIR>`) y ficheros de la carpeta en la que estamos.
* `cd ..`: subimos a la carpeta inmediatamente superior.
* `cd \\` (`cd \`): subimos a la carpeta raíz.
* `cd nombre_de_carpeta`: nos movemos a otra carpeta de nombre `nombre_de_carpeta`.
* `mkdir dir1` (`md dir1`): crea una carpeta de nombre `dir1` en la carpeta en la que nos encontremos.
* `rmdir dir1` (`rm dir1`): elimina la carpeta de nombre `dir1` (la eliminación se realiza si la carpeta a borrar está vacía).
* `del nombre_fichero`: elimina el fichero de nombre `nombre_fichero`.
* `more fichero.txt`: muestra el contenido de un fichero.
* `python`: arranca el intérprete de python. Para salir hay que teclear `quit()`.

## Primer programa en python

Podemos crear y ejecutar un programa en python del siguiente modo:
* Abrir el program `Anaconda CMD.exe Prompt`.
* Ejecutar la orden `notepad`, `notepad++` o `code` y crear un fichero de nombre `prueba1.py`.
* Escribir en el fichero `prueba1.py` lo siguiente:
<pre>
cad=input('Dime tu nombre:')
print('Hola',cad)
</pre>
* Guardar el fichero `prueba1.py`
* Ejecutar la orden `python.exe prueba1.py` en el `Anaconda Command Prompt`.

Definición de la función `factorial`

```python
import sys
sys.setrecursionlimit(1000)
def factorial(n):
    if n==0:
        return 1
    else:
        return factorial(n-1)*n
```

```python
x=factorial(500)
print( str(type( x )) + "\n" + str(x))
```
