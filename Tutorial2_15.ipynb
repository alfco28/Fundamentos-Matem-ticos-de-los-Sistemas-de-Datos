Tenemos tres formas de abrir un fichero de texto: 'r'
(modo lectura), 'w' (modo escritura), 'a' (modo añadir).
La orden para abrir un fichero es `open` y su
sintaxis es
> `file = open( ruta_hasta_el_fichero, modo_apertura)`

Los modos de apertura hacen lo siguiente:
* 'r': Abre el fichero para lectura y sitúa la posición de lectura al comienzo del mismo. El fichero ha de existir.
* 'w': Abre el fichero para escritura y sitúa la posición de escritura al comienzo del mismo. Si existe el fichero lo sobreescribe. Si el fichero no existe lo crea.
* 'a': Abre el fichero para escritura y sitúa la posición de escritura al final del mismo. Si el fichero no existe lo crea.

Para escribir en un fichero tenemos la orden `write`:
> `file.write('hola')`

Para la lectura tenemos `read()` que lee el fichero completo:
> `cadena=file.read()`

Por último, al finalizar de trabajar con un fichero debemos cerrarlo con la orden `close`:
> `file.close()`

## Librería `json`

## Resumen de la Librería `json` en Python

La librería **`json`** es un módulo nativo de Python que permite codificar y decodificar datos en formato JSON (JavaScript Object Notation). Es una herramienta fundamental para compartir datos entre aplicaciones, consumir APIs web y guardar configuraciones.

---

### ⚙️ Operaciones Principales

La librería se divide principalmente en dos tipos de operaciones: **Serialización** (convertir objetos de Python a JSON) y **Deserialización** (convertir JSON a objetos de Python).

#### 1. Serialización (Python ➔ JSON)
Para convertir objetos de Python (diccionarios, listas, etc.) en formato JSON.

*   `json.dumps(obj)`: Convierte un objeto de Python en una **cadena de texto (string)** en formato JSON.
*   `json.dump(obj, file_object)`: Convierte un objeto de Python y lo escribe directamente en un **archivo**.

#### 2. Deserialización (JSON ➔ Python)
Para transformar datos en formato JSON de vuelta a estructuras de datos de Python.

*   `json.loads(string)`: Modifica una **cadena de texto** JSON y la transforma en un objeto de Python (generalmente un diccionario o lista).
*   `json.load(file_object)`: Lee un **archivo** JSON y lo transforma en un objeto de Python.

---

### 📊 Tabla de Conversión de Datos

Python y JSON manejan los tipos de datos de forma equivalente pero con sutiles diferencias de sintaxis:

| Objeto Python | Equivalente JSON | Ejemplo Python ➔ JSON |
| :--- | :--- | :--- |
| `dict` | Object | `{"clave": "valor"}` |
| `list`, `tuple` | Array | `["uno", "dos"]` |
| `str` | String | `"Texto"` |
| `int`, `float` | Number | `42` o `3.14` |
| `True` / `False` | `true` / `false` | `true` / `false` (en minúscula) |
| `None` | `null` | `null` |

---

### 💻 Ejemplos de Código

#### Escribir y Leer una Cadena de Texto (Strings)
```python
import json

# Datos en Python
usuario = {
    "nombre": "Carlos",
    "edad": 30,
    "es_administrador": False,
    "hobbies": ["programar", "leer"]
}

# 1. Serializar: Python a String JSON
json_string = json.dumps(usuario, indent=4)
print(json_string)

# 2. Deserializar: String JSON a Python
datos_usuario = json.loads(json_string)
print(datos_usuario["nombre"])  # Salida: Carlos
```

#### Trabajar con Archivos (`.json`)
```python
import json

data = {"proyecto": "Alfa", "version": 1.0}

# Guardar en un archivo
with open("config.json", "w") as archivo:
    json.dump(data, archivo, indent=4)

# Leer desde el archivo
with open("config.json", "r") as archivo:
    datos_importados = json.load(archivo)
    
print(datos_importados)
```

---

### 🛠️ Parámetros Útiles

Al usar `dumps()` o `dump()`, puedes mejorar la legibilidad o estructura del JSON resultante:

*   **`indent`**: Define el número de espacios para la sangría. Ideal para que los humanos puedan leer el archivo (`indent=4`).
*   **`sort_keys`**: Si se define como `True`, ordena las claves del diccionario alfabéticamente.
*   **`ensure_ascii`**: Por defecto es `True` y escapa caracteres no ASCII (como las tildes o la ñ). Si se cambia a `False`, exportará los caracteres tal cual.

````python
import json
import random
from sys import argv

def creaFichero():
    try:
        N=int(10) # N=int(argv[1])
        nf="a.txt" #nf=argv[2]
        if N>1000 or N<=0:
            print("tienes que dar un valor positivo menor que 1000")
            return 
    except Exception as e:
        print(e)
        print("Ejemplo de llamada al programa:")
        print("  > python creafichero.py 10 a.txt")
    
    l1=[random.randint(0,100) for x in range(N)]
    cad=json.dumps(l1)
    
    try:
        fich=open(nf,"w")
        fich.write(cad)
        fich.close()
    except Exception as e:
        print(f"Error al crear el fichero {nf}")

if __name__ == '__main__':
    creaFichero()
````

````python
import struct, random
lfloat=[random.random() for x in range(5)]
print(lfloat)
lbytes=struct.pack( "%if" % len(lfloat), *lfloat )
print( type( lbytes), len(lbytes) )
struct.unpack( "%if" % len(lfloat), lbytes )
````

Indicamos con una cadena, por ejemplo "5d",  el formato y la cantidad de datos que queremos pasar a binario (en nuestor ejemplo cinco valores de tipo `double`)

````python
import struct, random
lfloat=[random.random() for x in range(5)]
print(lfloat)
lbytes=struct.pack( "{}d".format( len(lfloat) ), *lfloat )
print( type( lbytes), len(lbytes) )
struct.unpack( "{}d".format(len(lfloat)), lbytes )
````

## Librería `requests`

## Resumen de la Librería `requests` en Python

La librería **`requests`** es una biblioteca de terceros estándar de facto para realizar peticiones HTTP en Python. Diseñada bajo el lema *"HTTP for Humans"*, destaca por su sintaxis extremadamente limpia, intuitiva y por abstraer la complejidad técnica de bajo nivel.

---

### ⚙️ Métodos HTTP Principales

Permite interactuar con recursos web utilizando los verbos estándar del protocolo HTTP.

*   `requests.get(url)`: Solicita y descarga información de una URL.
*   `requests.post(url, data=...)`: Envía datos a un servidor (frecuente en formularios o APIs).
*   `requests.put(url, data=...)`: Reemplaza o actualiza por completo un recurso en el servidor.
*   `requests.delete(url)`: Borra un recurso específico del servidor.

---

### 📦 Anatomía de la Respuesta (`Response`)

Al ejecutar una petición, se genera un objeto de respuesta con propiedades clave pre-procesadas:

| Propiedad / Método | Descripción | Ejemplo de Uso |
| :--- | :--- | :--- |
| `.status_code` | Código numérico del estado HTTP | `200` (OK), `404` (Not Found) |
| `.text` | Contenido de la respuesta en **texto (str)** | Devuelve el HTML o texto plano decodificado |
| `.content` | Contenido de la respuesta en **bytes raw** | Ideal para descargar imágenes, PDFs o archivos |
| `.json()` | Convierte la respuesta JSON directamente | Transforma el cuerpo en un `dict` o `list` de Python |
| `.headers` | Diccionario con las cabeceras del servidor | Acceso a cookies, tipo de contenido, etc. |

---

### 💻 Ejemplos de Código

#### Petición GET Básica y Lectura de JSON
```python
import requests

# Consumir una API pública
url = "https://github.com"
response = requests.get(url)

if response.status_code == 200:
    # El método .json() ahorra importar la librería json y usar .loads()
    datos = response.json()
    print(f"Usuario: {datos['login']} - Repositorios: {datos['public_repos']}")
```

#### Petición POST Enviando Datos JSON y Cabeceras
```python

url = "https://httpbin.org"
cabeceras = {"User-Agent": "MiScriptPython/1.0"}
```

*   **`raise_for_status()`**: Método útil que lanza automáticamente una excepción si la petición devuelve un código de error (como un 400 o 500).

````python
import requests
def obtenerDatosDeIndice(indice,anioI,mesI,anioF,mesF,acumularEnTrimestres=False):
    '''
    Obtiene una lista de valores para un índice dado entre dos fechas. 
    Si acumularEnTrimestres es True acumula los valores por trimestre
    anioI, mesI: año y mes inicio
    anioF, mesF: año y mes final
    '''
    if mesI<10:
        mesI='0'+str(mesI)
    if mesF<10:
        mesF='0'+str(mesF)
    direccion=crearURL('ES','DATOS_SERIE',indice,'date={}{}01:{}{}31'.format(anioI,mesI,anioF,mesF)) #'nult=3')
    print("dirección a la que me conecto:",direccion)
    cnx=requests.get(direccion)
    diccionario=cnx.json()
    cnx.close()
    # Vemas la diferencia en segundos entre las dos primeras fechas
    deltaFechas=(diccionario['Data'][1]['Fecha']-diccionario['Data'][0]['Fecha'])/1000
    daux=[x['Valor'] for x in diccionario['Data']]
    # sólo podemos acumular en trimestres si los datos son mesuales
    # deltaFechas <= 3600*24*31 ocurre si los datos son mensuales
    if acumularEnTrimestres and deltaFechas <= 3600*24*31:
        daux = [sum(daux[i:i+3]) for i in range(0,len(daux),3)]
    return daux
````

### 2. Estructura Básica de la Petición
La URL base del recurso OpenAPI proporcionado es: `https://juntadeandalucia.es`. 

Sin embargo, para consultar el catálogo o listado real de las bibliotecas, debemos apuntar al endpoint de datos correspondiente (normalmente documentado dentro del propio JSON de la OpenAPI o derivado de la estructura `/api/v0/libraries`).

Aquí tienes el patrón estándar para realizar una petición `GET`, comprobar que el servidor responde correctamente (Código de estado `200 OK`) y transformar los datos:

```python
import requests

# 1. Definir la URL del recurso/endpoint de la API
url_api = "https://juntadeandalucia.es" # Ajustar según el endpoint específico de registros

try:
    # 2. Realizar la petición GET al servidor
    response = requests.get(url_api, timeout=10)
    
    # 3. Verificar si la petición fue exitosa (Status Code 200)
    if response.status_code == 200:
        # 4. Parsear el contenido de la respuesta directamente a un diccionario/lista de Python
        datos_bibliotecas = response.json()
        print(f"¡Conexión exitosa! Se han recuperado datos.")
        
        # Ejemplo: Mostrar la estructura de los primeros registros
        # print(datos_bibliotecas[:2]) 
        
    else:
        print(f"Error en la petición. Código de estado HTTP: {response.status_code}")

except requests.exceptions.Timeout:
    print("Error: Tiempo de espera agotado al conectar con el servidor.")
except requests.exceptions.ConnectionError:
    print("Error: No se pudo establecer la conexión a Internet.")
except Exception as e:
    print(f"Ocurrió un error inesperado: {e}")
```

### 3. Flujo para la Práctica (Guardado Local en JSON)
Una vez que la variable `datos_bibliotecas` contiene la información en formato nativo de Python, el proceso para cumplir con los requisitos de almacenamiento de la asignatura consiste en volcar este objeto a un archivo de texto plano utilizando el módulo estándar `json`:

```python
import json

# Suponiendo que 'datos_bibliotecas' ya contiene la información parseada de la API:
fichero_local = "mis_bibliotecas.json"

with open(fichero_local, "w", encoding="utf-8") as f:
    json.dump(datos_bibliotecas, f, ensure_ascii=False, indent=4)

print(f"Datos respaldados localmente en '{fichero_local}'")
```
