### Tipo `int`

Los operadores para el tipo de dato `int` en Python se dividen entre **unarios** (requieren un solo operando) y **binarios** (requieren dos operandos). Se clasifican en categorías aritméticas, de comparación y a nivel de bits (*bitwise*).

#### 1. Operadores Unarios
Actúan sobre un único número entero para cambiar su signo o invertir sus bits:

| Operador | Operación | Ejemplo (`a = 5` [bin: `0101`]) | Resultado |
| :--- | :--- | :--- | :--- |
| `+` | **Identidad** (Mantiene el signo original) | `+a` | `5` |
| `-` | **Negación aritmética** (Cambia el signo) | `-a` | `-5` |
| `~` | **Inversión bit a bit** (Complemento a uno: `-(x + 1)`) | `~a` (da `...1010`) | `-6` |

#### 2. Operadores Binarios Aritméticos
Permiten realizar operaciones matemáticas estándar entre dos números enteros:

| Operador | Operación | Ejemplo (`a = 10`, `b = 3`) | Resultado |
| :--- | :--- | :--- | :--- |
| `+` | **Suma** | `a + b` | `13` |
| `-` | **Resta** | `a - b` | `7` |
| `*` | **Multiplicación** | `a * b` | `30` |
| `/` | **División flotante** | `a / b` | `3.3333...` (Devuelve un `float`) |
| `//` | **División entera** (cociente) | `a // b` | `3` |
| `%` | **Módulo** (resto de la división) | `a % b` | `1` |
| `**` | **Exponente** (potencia) | `a ** b` | `1000` |

#### 3. Operadores Binarios de Comparación (Relacionales)
Comparan dos enteros y devuelven un valor booleano (`True` o `False`):

| Operador | Operación | Ejemplo (`a = 10`, `b = 3`) | Resultado |
| :--- | :--- | :--- | :--- |
| `==` | **Igual a** | `a == b` | `False` |
| `!=` | **Diferente de** | `a != b` | `True` |
| `>` | **Mayor que** | `a > b` | `True` |
| `<` | **Menor que** | `a < b` | `False` |
| `>=` | **Mayor o igual que** | `a >= b` | `True` |
| `<=` | **Menor o igual que** | `a <= b` | `False` |

#### 4. Operadores Binarios a Nivel de Bits (*Bitwise*)
Operan directamente sobre la **representación binaria** de dos números enteros:

| Operador | Operación | Ejemplo (`a = 5` [bin: `0101`], `b = 3` [bin: `0011`]) | Resultado |
| :--- | :--- | :--- | :--- |
| `&` | **AND** bit a bit | `a & b` (da `0001`) | `1` |
| `\|` | **OR** bit a bit | `a \| b` (da `0111`) | `7` |
| `^` | **XOR** bit a bit | `a ^ b` (da `0110`) | `6` |
| `<<` | **Desplazamiento a la izquierda** | `a << 1` (da `1010`) | `10` |
| `>>` | **Desplazamiento a la derecha** | `a >> 1` (da `0010`) | `2` |

### Tipo `float`

Los operadores para el tipo de dato `float` en Python se dividen entre **unarios** (requieren un solo operando) y **binarios** (requieren dos operandos). Al igual que con los tipos enteros, los flotantes **no son compatibles con las operaciones a nivel de bits** (como `~`, `&`, `|`, `^`, `<<`, `>>`).

#### 1. Operadores Unarios
Actúan sobre un único número flotante para alterar o mantener su signo:

| Operador | Operación | Ejemplo (`a = 5.5`) | Resultado |
| :--- | :--- | :--- | :--- |
| `+` | **Identidad** (Mantiene el signo original) | `+a` | `5.5` |
| `-` | **Negación aritmética** (Invierte el signo) | `-a` | `-5.5` |

#### 2. Operadores Binarios Aritméticos
Realizan operaciones matemáticas tradicionales. Al operar con al menos un `float`, el resultado devuelto siempre será de tipo `float`:

| Operador | Operación | Ejemplo (`a = 10.5`, `b = 2.0`) | Resultado |
| :--- | :--- | :--- | :--- |
| `+` | **Suma** | `a + b` | `12.5` |
| `-` | **Resta** | `a - b` | `8.5` |
| `*` | **Multiplicación** | `a * b` | `21.0` |
| `/` | **División flotante** | `a / b` | `5.25` |
| `//` | **División entera** (truncada) | `a // b` | `5.0` (Resultado flotante con decimal `.0`) |
| `%` | **Módulo** (resto) | `a % b` | `0.5` |
| `**` | **Exponente** (potencia) | `a ** b` | `110.25` |

#### 3. Operadores Binarios de Comparación (Relacionales)
Comparan dos números y devuelven un valor booleano (`True` o `False`):

| Operador | Operación | Ejemplo (`a = 10.5`, `b = 2.0`) | Resultado |
| :--- | :--- | :--- | :--- |
| `==` | **Igual a** | `a == b` | `False` |
| `!=` | **Diferente de** | `a != b` | `True` |
| `>` | **Mayor que** | `a > b` | `True` |
| `<` | **Menor que** | `a < b` | `False` |
| `>=` | **Mayor o igual que** | `a >= b` | `True` |
| `<=` | **Menor o igual que** | `a <= b` | `False` |

*(Nota: Evita usar la igualdad exacta `==` para flujos lógicos críticos debido a las limitaciones de precisión de la norma IEEE 754 de punto flotante).*

### Tipo `complex`

El tipo de dato <code>complex</code> maneja numeros complejos.
Definimos <code>a=5.1+7.34j</code> y vemos su tipo

Los operadores para el tipo de dato `complex` en Python se dividen entre **unarios** (requieren un solo operando) y **binarios** (requieren dos operandos). 

Los números complejos **no son compatibles con las operaciones de orden** (`<`, `>`, `<=`, `>=`), división entera (`//`), módulo (`%`), ni operaciones a nivel de bits (*bitwise*).

#### 1. Operadores Unarios
Actúan sobre un único número complejo para alterar o mantener el signo de sus componentes (tanto la parte real como la imaginaria):

| Operador | Operación | Ejemplo (`a = 4+2j`) | Resultado |
| :--- | :--- | :--- | :--- |
| `+` | **Identidad** (Mantiene los signos originales) | `+a` | `(4+2j)` |
| `-` | **Negación aritmética** (Invierte el signo de ambas partes) | `-a` | `(-4-2j)` |

#### 2. Operadores Binarios Aritméticos
Realizan operaciones algebraicas entre números complejos. Al operar con un `complex`, el resultado final siempre es de tipo `complex`:

| Operador | Operación | Ejemplo (`a = 4+2j`, `b = 1+1j`) | Resultado |
| :--- | :--- | :--- | :--- |
| `+` | **Suma** | `a + b` | `(5+3j)` |
| `-` | **Resta** | `a - b` | `(3+1j)` |
| `*` | **Multiplicación** | `a * b` | `(2+6j)` |
| `/` | **División** | `a / b` | `(3-1j)` |
| `**` | **Exponente** (potencia) | `a ** 2` | `(12+16j)` |

#### 3. Operadores Binarios de Comparación (Relacionales)
Debido a que los números complejos representan puntos en un plano bidimensional y no tienen un orden lineal, **solo se pueden comparar por igualdad exacta**:

| Operador | Operación | Ejemplo (`a = 4+2j`, `b = 1+1j`) | Resultado |
| :--- | :--- | :--- | :--- |
| `==` | **Igual a** | `a == b` | `False` |
| `!=` | **Diferente de** | `a != b` | `True` |

*(Nota: El uso de `<`, `>`, `<=`, o `>=` con números complejos arrojará un error de tipo `TypeError`).*

### Tipo `bool`

Los operadores para el tipo de dato `bool` en Python se dividen entre **unarios** (un solo operando) y **binarios** (dos operandos). 

Dado que en Python `bool` es una subclase de `int` (donde `True` equivale a `1` y `False` a `0`), este tipo de dato hereda capacidades matemáticas y de manipulación de bits, además de sus operadores lógicos nativos.

#### 1. Operadores Unarios
Actúan sobre un único valor booleano. Incluye el operador lógico de negación y los operadores heredados de los enteros:

| Operador | Operación | Ejemplo (`a = True`) | Resultado |
| :--- | :--- | :--- | :--- |
| `not` | **Negación lógica** (Invierte el valor lógico) | `not a` | `False` |
| `+` | **Identidad aritmética** (Convierte a entero `1` o `0`) | `+a` | `1` (Tipo `int`) |
| `-` | **Negación aritmética** (Convierte a entero con signo invertido) | `-a` | `-1` (Tipo `int`) |
| `~` | **Inversión bit a bit** (Complemento a uno: `-(x + 1)`) | `~a` | `-2` (Tipo `int`) |

#### 2. Operadores Binarios Lógicos
Son los operadores nativos principales para evaluar expresiones de lógica booleana con cortocircuito (evaluación perezosa):

| Operador | Operación | Ejemplo (`a = True`, `b = False`) | Resultado |
| :--- | :--- | :--- | :--- |
| `and` | **Y lógico** (Verdadero si ambos son True) | `a and b` | `False` |
| `or` | **O lógico** (Verdadero si al menos uno es True) | `a or b` | `True` |

#### 3. Operadores Binarios de Comparación (Relacionales)
Comparan los valores booleanos basándose en su equivalencia numérica (`False < True` porque `0 < 1`):

| Operador | Operación | Ejemplo (`a = True`, `b = False`) | Resultado |
| :--- | :--- | :--- | :--- |
| `==` | **Igual a** | `a == b` | `False` |
| `!=` | **Diferente de** (Funciona como un XOR lógico) | `a != b` | `True` |
| `>` | **Mayor que** | `a > b` | `True` |
| `<` | **Menor que** | `a < b` | `False` |
| `>=` | **Mayor o igual que** | `a >= b` | `True` |
| `<=` | **Menor o igual que** | `a <= b` | `False` |

#### 4. Operadores Binarios a Nivel de Bits (*Bitwise*)
Evalúan ambos operandos obligatoriamente (sin cortocircuito) y devuelven un objeto de tipo `bool`:

| Operador | Operación | Ejemplo (`a = True`, `b = False`) | Resultado |
| :--- | :--- | :--- | :--- |
| `&` | **AND** bit a bit | `a & b` | `False` |
| `\|` | **OR** bit a bit | `a \| b` | `True` |
| `^` | **XOR** bit a bit | `a ^ b` | `True` |

#### 5. Operadores Binarios Aritméticos
Fuerzan la conversión implícita (*coerción*) de los booleanos a enteros (`1` y `0`) para realizar cálculos matemáticos:

| Operador | Operación | Ejemplo (`a = True`, `b = False`) | Resultado |
| :--- | :--- | :--- | :--- |
| `+` | **Suma** | `a + a` | `2` (Tipo `int`) |
| `-` | **Resta** | `a - b` | `1` (Tipo `int`) |
| `*` | **Multiplicación** | `a * b` | `0` (Tipo `int`) |
| `/` | **División** | `b / a` | `0.0` (Tipo `float`) |

## Cadenas de caracteres

Las cadenas de caracteres son un tipo de dato bastante usado.
Con la orden <code>c='aaa\n\tbbb\n'</code> definimos una cadena
de caracteres.

Los operadores y funciones para el tipo de dato `str` en Python permiten manipular, buscar, validar y transformar secuencias de texto. A diferencia de los tipos numéricos, las cadenas son **inmutables**, lo que significa que todas las funciones que modifican el texto devuelven una nueva cadena.

### 1. Operadores Unarios y Binarios para `str`
Los operadores en cadenas se utilizan principalmente para concatenación, repetición, validación de pertenencia e identidad:

| Operador | Tipo | Operación | Ejemplo (`a = "Hola"`, `b = "Mundo"`) | Resultado |
| :--- | :--- | :--- | :--- | :--- |
| `+` | Binario | **Concatenación** (Une dos cadenas) | `a + " " + b` | `"Hola Mundo"` |
| `*` | Binario | **Repetición** (Multiplica por un `int`) | `a * 3` | `"HolaHolaHola"` |
| `in` | Binario | **Pertenencia** (Verifica si una subcadena existe) | `"ol" in a` | `True` |
| `not in`| Binario | **Ausencia** (Verifica si una subcadena no existe) | `"x" not in a` | `True` |
| `==` | Binario | **Igualdad exacta** (Sensible a mayúsculas) | `a == "hola"` | `False` |
| `!=` | Binario | **Diferencia** | `a != b` | `True` |
| `<`, `>` | Binario | **Comparación lexicográfica** (Orden ASCII) | `"Ana" < "Juan"` | `True` |

### 2. Funciones Integradas (*Built-in Functions*)
Funciones globales de Python que aceptan cadenas como argumento:

| Función | Descripción | Ejemplo | Resultado |
| :--- | :--- | :--- | :--- |
| `len(s)` | Devuelve la longitud (número de caracteres). | `len("Python")` | `6` |
| `str(x)` | Convierte cualquier objeto `x` a tipo cadena. | `str(123)` | `"123"` |
| `max(s)` | Devuelve el carácter con el mayor valor ASCII. | `max("abc")` | `"c"` |
| `min(s)` | Devuelve el carácter con el menor valor ASCII. | `min("abc")` | `"a"` |

### 3. Métodos de Formato y Transformación (`str.method()`)
Métodos que transforman el texto o cambian su estructura visual:

| Método | Descripción | Ejemplo (`s = " hola mundo "`) | Resultado |
| :--- | :--- | :--- | :--- |
| `.upper()` | Convierte todo el texto a mayúsculas. | `"Python".upper()` | `"PYTHON"` |
| `.lower()` | Convierte todo el texto a minúsculas. | `"Python".lower()` | `"python"` |
| `.capitalize()` | Coloca la primera letra en mayúscula y el resto en minúscula. | `"hola".capitalize()` | `"Hola"` |
| `.strip()` | Elimina los espacios en blanco al inicio y al final. | `s.strip()` | `"hola mundo"` |
| `.replace(v, n)`| Reemplaza el valor viejo `v` por el nuevo `n`. | `"casa".replace("a", "e")` | `"cese"` |
| `.split(sep)` | Divide la cadena en una lista usando un separador. | `"a,b,c".split(",")` | `['a', 'b', 'c']` |
| `.join(iterable)`| Une elementos de una lista/colección usando la cadena como nexo. | `"-".join(["A", "B"])` | `"A-B"` |

### 4. Métodos de Búsqueda y Validación
Métodos empleados para inspeccionar el contenido de la cadena:

| Método | Descripción | Ejemplo (`s = "Python"`) | Resultado |
| :--- | :--- | :--- | :--- |
| `.find(sub)` | Busca una subcadena; devuelve el índice de la primera coincidencia (o `-1` si no existe). | `"hola".find("o")` | `1` |
| `.count(sub)` | Cuenta cuántas veces aparece una subcadena sin superponerse. | `"banana".count("an")` | `2` |
| `.startswith(p)`| Verifica si la cadena comienza con el prefijo `p`. | `s.startswith("Py")` | `True` |
| `.endswith(s)` | Verifica si la cadena termina con el sufijo `s`. | `s.endswith("on")` | `True` |
| `.isdigit()` | Retorna `True` si todos los caracteres son números enteros. | `"123".isdigit()` | `True` |
| `.isalpha()` | Retorna `True` si todos los caracteres son alfabéticos (letras). | `"Py3".isalpha()` | `False` |

### Uso de `format`

Uso de `format`. Con esta orden podemos insertar los valores de varias variables dentro de una cadena de carateres. 
Se puede conseguir el mismo resultado concatenando cadenas con la orden `str`.

Las **f-strings** evalúan las expresiones dentro de las llaves `{}` en tiempo de ejecución, convirtiendo los resultados automáticamente a cadenas de texto.

#### 1. Sintaxis Básica e Interpolación
Permiten incrustar variables, operaciones matemáticas y llamadas a funciones o métodos directamente en el texto:

```python
nombre = "Ana"
edad = 28

# Interpolación de variables
print(f"Hola, mi nombre es {nombre} y tengo {edad} años.")
# Salida: Hola, mi nombre es Ana y tengo 28 años.

# Operaciones aritméticas y métodos
print(f"El próximo año tendré {edad + 1} años y mi nombre en mayúsculas es {nombre.upper()}.")
# Salida: El próximo año tendré 29 años y mi nombre en mayúsculas es ANA.
```

#### 2. Formateo de Números y Decimales
Utilizando los dos puntos `:` dentro de las llaves, puedes controlar la precisión de los flotantes o añadir separadores de miles:

| Sintaxis | Descripción | Ejemplo | Resultado |
| :--- | :--- | :--- | :--- |
| `:{total_caracteres}.{decimales}f` | Limita los decimales de un `float`. | `f"{3.14159:.2f}"` | `"3.14"` |
| `:,` | Añade comas como separador de miles. | `f"{1000000:,}"` | `"1,000,000"` |
| `:.2%` | Convierte un número a formato porcentaje. | `f"{0.75:-.1%}"` | `"75.0%"` |

#### 3. Alineación y Relleno de Texto
Puedes rellenar con espacios o caracteres específicos y alinear el texto a la izquierda, derecha o centro:

| Sintaxis | Descripción | Ejemplo | Resultado |
| :--- | :--- | :--- | :--- |
| `:<N` | Alinea a la izquierda (en un ancho de N). | `f"{'Ok':<5}"` | `"Ok   "` |
| `:>N` | Alinea a la derecha. | `f"{'Ok':>5}"` | `"   Ok"` |
| `:^N` | Centra el texto. | `f"{'Ok':^5}"` | `"  Ok "` |
| `:{carácter}{alineación}{N}` | Rellena con un carácter personalizado. | `f"{'Ok':*^6}"` | `"**Ok**"` |

#### 4. Formateo de Fechas (Datetime)
Extrae componentes específicos de un objeto de tipo fecha directamente sin llamar a `strftime`:

```python
from datetime import datetime
ahora = datetime.now()

print(f"Hoy es {ahora:%A} del mes {ahora:%B}.")
# Salida: Hoy es Monday del mes September. (Depende del idioma del sistema)
```

#### 5. Modo Depuración (Python 3.8+)
Si añades un signo de igual `=` después de la variable o expresión, Python imprimirá tanto la expresión literal como su resultado actual. Es ideal para hacer `print()` rápidos:

```python
x = 10
y = 25

print(f"{x=}, {y=}, {x+y=}")
# Salida: x=10, y=25, x+y=35
```

#### ⚠️ Reglas Importantes de Sintaxis:
1. **Comillas internas:** Si la f-string usa comillas dobles externas (`f"..."`), las llaves internas para los diccionarios deben usar comillas simples para no romper la cadena: `f"Usuario: {datos['nombre']}"`.
2. **No usar barras invertidas (`\`) dentro de `{}`:** No puedes meter saltos de línea `\n` o caracteres de escape dentro de las expresiones evaluadas. Haz la operación fuera de la llave o de la f-string.
3. **Escapar llaves:** Si necesitas que aparezcan llaves literales `{}` impresas en el texto final, debes duplicarlas: `f"Las llaves se escriben así: {{variable}}"` imprimirá `Las llaves se escriben así: {variable}`.

# Colecciones

Colecciones: hay tres tipos de colecciones: listas, tuplas, diccionarios e iteradores

## Listas

Las **listas** son una de las estructuras de datos más utilizadas en Python debido a su flexibilidad y a la gran variedad de métodos integrados que ofrecen.

### 1. Operadores Básicos para Listas
Permiten combinar, duplicar y validar la pertenencia de elementos dentro de una lista:

| Operador | Tipo | Operación | Ejemplo (`a = [1, 2]`, `b = [3]`) | Resultado |
| :--- | :--- | :--- | :--- | :--- |
| `+` | Binario | **Concatenación** (Une dos listas) | `a + b` | `[1, 2, 3]` |
| `*` | Binario | **Repetición** (Multiplica por un `int`) | `a * 3` | `[1, 2, 1, 2, 1, 2]` |
| `in` | Binario | **Pertenencia** (Verifica si existe el ítem) | `2 in a` | `True` |
| `not in`| Binario | **Ausencia** (Verifica si no existe el ítem)| `5 not in a` | `True` |
| `==` | Binario | **Igualdad** (Mismos ítems y mismo orden) | `a == [1, 2]` | `True` |

### 2. Acceso y Slicing (Troceado)
Las listas utilizan índices basados en cero (`0`) para acceder a sus elementos:

```python
lista = ['a', 'b', 'c', 'd', 'e']

# Indexación básica
print(lista[0])   # 'a' (Primer elemento)
print(lista[-1])  # 'e' (Último elemento usando índices negativos)

# Slicing: lista[inicio:fin:paso] (El índice 'fin' queda excluido)
print(lista[1:4])   # ['b', 'c', 'd']
print(lista[:3])    # ['a', 'b', 'c'] (Desde el inicio)
print(lista[::2])   # ['a', 'c', 'e'] (Saltando de 2 en 2)
print(lista[::-1])  # ['e', 'd', 'c', 'b', 'a'] (Invierte la lista)
```

### 3. Métodos para Modificar Listas (`list.method()`)
Como las listas son **mutables**, estos métodos modifican la lista original directamente (*in-place*):

| Método | Descripción | Ejemplo (`L = [1, 2, 3]`) | Resultado en `L` |
| :--- | :--- | :--- | :--- |
| `.append(x)` | Añade el elemento `x` al final de la lista. | `L.append(4)` | `[1, 2, 3, 4]` |
| `.extend(iter)`| Agrega los elementos de otro iterable al final. | `L.extend([4, 5])` | `[1, 2, 3, 4, 5]` |
| `.insert(i, x)`| Inserta el elemento `x` en la posición del índice `i`. | `L.insert(1, 9)` | `[1, 9, 2, 3]` |
| `.remove(x)` | Elimina la **primera** aparición del elemento `x`. | `L.remove(2)` | `[1, 3]` (Lanza error si no existe) |
| `.pop(i)` | Elimina y **retorna** el elemento en el índice `i` (por defecto el último). | `L.pop(0)` | Devuelve `1`, la lista queda `[2, 3]` |
| `.clear()` | Vacía por completo la lista. | `L.clear()` | `[]` |
| `.sort()` | Ordena la lista (ascendente de forma nativa). | `[3, 1, 2].sort()`| Modifica la lista a `[1, 2, 3]` |
| `.reverse()` | Invierte el orden de los elementos actuales. | `L.reverse()` | `[3, 2, 1]` |

### 4. Funciones Integradas Útiles
Funciones globales de Python que aceptan listas como argumento:

| Función | Descripción | Ejemplo (`L = [10, 20, 30]`) | Resultado |
| :--- | :--- | :--- | :--- |
| `len(L)` | Devuelve la cantidad de elementos. | `len(L)` | `3` |
| `max(L)` / `min(L)` | Devuelve el valor máximo / mínimo. | `max(L)` | `30` |
| `sum(L)` | Suma todos los elementos numéricos. | `sum(L)` | `60` |
| `sorted(L)` | Devuelve una **nueva** lista ordenada sin alterar la original. | `sorted([3, 1])` | `[1, 3]` |

### 5. Comprensión de Listas (*List Comprehensions*)
Es una sintaxis concisa y elegante para crear nuevas listas a partir de iterables existentes:

```python
# Sintaxis básica: [expresión for ítem in iterable si condición]
numeros = [1, 2, 3, 4, 5]

# Crear una lista con los cuadrados de los números pares
cuadrados_pares = [x**2 for x in numeros if x % 2 == 0]
# Resultado: [4, 16]
```

## Tuplas

Las tuplas son de tipo `tuple`.

No se pueden modificar y por ello se usan normalmente como argumentos a funciones.

Pueden convertirse a listas con la orden `list(tpl)`.

Para convertir una lista en una tupla tenemos la orden `tuple(l1)`.

Al igual que las listas pueden concatenarse y en ellas tenemos a las funciones `count` e `index`, además de `len`, `max`, `min`, `sum`, `sorted`, `any` y `all`.

También tenemos para ellas las funciones `*` y `+`

La tupla vacía es `()` y la tupla formada por un elemento `x` es `(x,)`.

Las **tuplas** se utilizan principalmente para almacenar datos que no deben cambiar a lo largo de la ejecución del programa. Al ser inmutables, son más rápidas que las listas y consumen menos memoria.

### 1. Creación y Reglas de Sintaxis
Puedes crear tuplas con o sin paréntesis. La presencia de la coma es lo que realmente define a una tupla:

```python
# Creación estándar
mi_tupla = (1, 2, 3)
otra_tupla = "a", "b", "c"  # Empaquetado de tuplas (sin paréntesis)

# ¡IMPORTANTE! Tupla de un solo elemento (requiere una coma al final)
tupla_unica = (5,)  # Si escribes (5), Python lo interpretará como un entero normal
```

### 2. Operadores Básicos para Tuplas
Al igual que las listas, permiten combinar, duplicar y validar la pertenencia de elementos:

| Operador | Tipo | Operación | Ejemplo (`a = (1, 2)`, `b = (3, 4)`) | Resultado |
| :--- | :--- | :--- | :--- | :--- |
| `+` | Binario | **Concatenación** (Crea una nueva tupla) | `a + b` | `(1, 2, 3, 4)` |
| `*` | Binario | **Repetición** | `a * 2` | `(1, 2, 1, 2)` |
| `in` | Binario | **Pertenencia** | `1 in a` | `True` |
| `not in`| Binario | **Ausencia** | `5 not in a` | `True` |
| `==` | Binario | **Igualdad** (Mismos ítems y mismo orden) | `a == (1, 2)` | `True` |

### 3. Acceso y Slicing (Troceado)
Las tuplas utilizan índices basados en cero (`0`) exactamente igual que las listas:

```python
frutas = ("manzana", "plátano", "naranja", "uva")

print(frutas[0])    # "manzana"
print(frutas[-1])   # "uva" (Índice negativo para el final)
print(frutas[1:3])  # ("plátano", "naranja") (El índice final queda excluido)
```

### 4. Métodos Integrados (`tuple.method()`)
Dado que son inmutables, las tuplas **solo tienen dos métodos**:

| Método | Descripción | Ejemplo (`t = (1, 2, 2, 3)`) | Resultado |
| :--- | :--- | :--- | :--- |
| `.count(x)` | Cuenta cuántas veces aparece el elemento `x`. | `t.count(2)` | `2` |
| `.index(x)` | Devuelve el primer índice donde se encuentra `x`. | `t.index(3)` | `3` (Lanza ValueError si no existe) |

### 5. Desempaquetado de Tuplas (*Tuple Unpacking*)
Permite asignar los elementos de una tupla directamente a variables individuales:

```python
persona = ("Carlos", 30, "Madrid")

# Desempaquetado básico
nombre, edad, ciudad = persona
print(nombre)  # Carlos

# Desempaquetado avanzado usando el operador asterisco (*)
numeros = (1, 2, 3, 4, 5)
primero, *en_medio, ultimo = numeros
print(primero)   # 1
print(en_medio)  # [2, 3, 4] (Se extrae como una lista)
print(ultimo)    # 5
```

### 💡 ¿Cuándo usar una Tupla en lugar de una Lista?
1. **Seguridad de datos:** Para proteger datos que no deben ser alterados (ej. coordenadas geográficas, configuraciones del sistema).
2. **Llaves de diccionarios:** Debido a su inmutabilidad, las tuplas pueden usarse como llaves en un diccionario (`dict`), mientras que las listas no.
3. **Retorno de funciones:** Cuando una función devuelve múltiples valores separados por comas, técnicamente está retornando una tupla.

## Diccionarios

Un diccionario es similar a una lista pero las posiciones de sus elementos
no son los valores `0,1,2,...,` sino que cada posición tiene
un nombre que usaremos para acceder.

Los **diccionarios** son una de las estructuras más potentes de Python. A partir de Python 3.7, preservan el **orden de inserción** de sus elementos. Sus llaves deben ser de un tipo de dato inmutable (como `str`, `int` o `tuple`).

### 1. Operadores Básicos para Diccionarios
Permiten validar la presencia de llaves o fusionar estructuras (fusionar disponible desde Python 3.9):

| Operador | Tipo | Operación | Ejemplo (`a = {'x': 1}`, `b = {'y': 2}`) | Resultado |
| :--- | :--- | :--- | :--- | :--- |
| `\|` | Binario | **Unión/Fusión** (Une dos diccionarios) | `a \| b` | `{'x': 1, 'y': 2}` |
| `in` | Binario | **Pertenencia** (Busca en las **llaves**) | `'x' in a` | `True` |
| `not in`| Binario | **Ausencia** (Busca en las **llaves**) | `'z' not in a` | `True` |
| `==` | Binario | **Igualdad** (Mismos pares independientemente del orden) | `a == {'x': 1}` | `True` |

### 2. Acceso y Modificación Básica
El acceso estándar se realiza mediante corchetes utilizando la llave:

```python
capitales = {"España": "Madrid", "Francia": "París"}

# Acceso directo
print(capitales["España"])  # "Madrid"
# print(capitales["Italia"]) # KeyError (Detiene el programa si la llave no existe)

# Añadir o modificar elementos
capitales["Italia"] = "Roma"      # Añade un nuevo par
capitales["Francia"] = "Francia"  # Modifica un valor existente
```

### 3. Métodos Esenciales (`dict.method()`)
Métodos integrados para leer, extraer o limpiar elementos de forma segura:

| Método | Descripción | Ejemplo (`d = {'a': 1, 'b': 2}`) | Resultado |
| :--- | :--- | :--- | :--- |
| `.get(llave, df)`| Retorna el valor de la llave. Si no existe, devuelve el valor por defecto `df` sin lanzar error. | `d.get('c', 0)` | `0` |
| `.keys()` | Devuelve una vista con todas las **llaves** del diccionario. | `d.keys()` | `dict_keys(['a', 'b'])` |
| `.values()` | Devuelve una vista con todos los **valores**. | `d.values()` | `dict_values([1, 2])` |
| `.items()` | Devuelve una vista con tuplas de pares `(llave, valor)`. | `d.items()` | `dict_items([('a', 1), ('b', 2)])` |
| `.pop(llave)` | Elimina la llave y **retorna** su valor asociado. | `d.pop('a')` | Devuelve `1`; `d` queda `{'b': 2}` |
| `.update(otro)`| Actualiza el diccionario agregando o modificando pares desde otro diccionario o iterable. | `d.update({'b': 3, 'c': 4})`| `d` queda `{'b': 3, 'c': 4}` |
| `.clear()` | Elimina todos los elementos del diccionario. | `d.clear()` | `{}` |

### 4. Iteración en Diccionarios
Formas comunes de recorrer un diccionario utilizando bucles `for`:

```python
edades = {"Ana": 25, "Luis": 32}

# Iterar solo por las llaves (comportamiento por defecto)
for nombre in edades:
    print(nombre)
```

## Conjuntos

En python tenemos también un tipo de dato para definir conjuntos

Los **conjuntos** son ideales para eliminar duplicados de secuencias y realizar operaciones matemáticas de teoría de conjuntos (como uniones, intersecciones y diferencias). Sus elementos deben ser inmutables (como `int`, `str` o `tuple`).

### 1. Operadores Binarios para Conjuntos (Operaciones Matemáticas)
Permiten realizar operaciones de álgebra de conjuntos de forma matemática y directa:

| Operador | Operación | Descripción | Ejemplo (`a = {1, 2}`, `b = {2, 3}`) | Resultado |
| :--- | :--- | :--- | :--- | :--- |
| `\|` | **Unión** | Elementos que están en cualquiera de los dos conjuntos. | `a \| b` | `{1, 2, 3}` |
| `&` | **Intersección** | Elementos presentes en ambos conjuntos. | `a & b` | `{2}` |
| `-` | **Diferencia** | Elementos que están en `a` pero no en `b`. | `a - b` | `{1}` |
| `^` | **Diferencia simétrica**| Elementos que están en uno u otro, pero no en ambos. | `a ^ b` | `{1, 3}` |

### 2. Operadores de Comparación y Pertenencia
Utilizados para verificar relaciones lógicas entre conjuntos o elementos:

| Operador | Operación | Descripción | Ejemplo (`a = {1, 2}`, `b = {1, 2, 3}`) | Resultado |
| :--- | :--- | :--- | :--- | :--- |
| `in` | **Pertenencia** | Verifica si un elemento está en el conjunto. | `1 in a` | `True` |
| `not in`| **Ausencia** | Verifica si un elemento no está. | `5 not in a` | `True` |
| `<=` | **Subconjunto** | Verifica si todos los elementos de `a` están en `b`. | `a <= b` | `True` |
| `>=` | **Superconjunto** | Verifica si `b` contiene todos los elementos de `a`. | `b >= a` | `True` |
| `==` | **Igualdad** | Tienen exactamente los mismos elementos. | `a == {2, 1}` | `True` |

### 3. Métodos Esenciales para Modificar y Consultar (`set.method()`)
Métodos para añadir, eliminar o comparar conjuntos de forma explícita:

| Método | Equivalente | Descripción | Ejemplo (`s = {1, 2}`) | Resultado en `s` |
| :--- | :--- | :--- | :--- | :--- |
| `.add(x)` | *N/A* | Añade el elemento `x` al conjunto. | `s.add(3)` | `{1, 2, 3}` |
| `.remove(x)` | *N/A* | Elimina `x`. Lanza error (`KeyError`) si no existe. | `s.remove(1)` | `{2}` |
| `.discard(x)`| *N/A* | Elimina `x`. **No** lanza error si el elemento no está. | `s.discard(9)` | `{1, 2}` |
| `.pop()` | *N/A* | Elimina y retorna un elemento **aleatorio** (está desordenado).| `s.pop()` | Retorna `1` o `2` |
| `.clear()` | *N/A* | Vacía por completo el conjunto. | `s.clear()` | `set()` |
| `.isdisjoint(o)`| *N/A* | Devuelve `True` si no tienen ningún elemento en común.| `{1}.isdisjoint({2})`| `True` |

*(Nota: Los métodos `.union()`, `.intersection()`, `.difference()` y `.symmetric_difference()` son equivalentes a los operadores `|`, `&`, `-` y `^` respectivamente, pero con la ventaja de que aceptan cualquier iterable como argumento, no solo otros conjuntos).*

### 4. Comprensión de Conjuntos (*Set Comprehensions*)
Permite construir conjuntos de forma compacta. Al igual que los diccionarios, utiliza llaves `{}`, pero sin el formato `llave: valor`:

```python
texto = "abracadabra"

# Crea un conjunto con los caracteres únicos que no sean la letra 'a'
caracteres_unicos = {letra for letra in texto if letra != 'a'}
# Resultado: {'b', 'r', 'c', 'd'} (el orden puede variar)
```

# Definición de funciones

## Estructura condicional
`if, if...else, if...elif...elif...else`

En Python, los bloques de código condicionales se definen mediante la **indentación** (sangría de 4 espacios), eliminando la necesidad de usar llaves `{}` o palabras clave de cierre como `endif`.

### 1. Estructura Básica y Sintaxis
Una estructura condicional evalúa expresiones de izquierda a derecha. En el momento en que una condición se cumple (`True`), se ejecuta su bloque y se ignoran las demás.

```python
edad = 20

if edad >= 18:
    print("Eres mayor de edad.")
```

### 2. Estructura Completa (`if`, `elif`, `else`)
* **`if`**: La condición inicial obligatoria.
* **`elif`** (abreviación de *else if*): Permite evaluar múltiples condiciones consecutivas de forma ordenada. Puedes usar tantos como necesites.
* **`else`**: Bloque opcional que se ejecuta únicamente si **ninguna** de las condiciones anteriores fue verdadera.

```python
nota = 85

if nota >= 90:
    print("Excelente")
elif nota >= 70:
    print("Aprobado")
else:
    print("Reprobado")
```

### 3. Operador Ternario (Condicionales en una sola línea)
Es una sintaxis compacta para asignar valores a una variable evaluando una condición en una única línea de código:

```python
# Sintaxis: valor_si_verdadero if condicion else valor_si_falso
edad = 16
estado = "Adulto" if edad >= 18 else "Menor"

print(estado)  # Salida: Menor
```

### 4. Condiciones Múltiples (Operadores Lógicos)
Puedes combinar varias evaluaciones dentro de un mismo `if` utilizando los operadores lógicos `and`, `or` y `not`:

```python
ingresos = 2500
tiene_deudas = False

if ingresos > 2000 and not tiene_deudas:
    print("Crédito aprobado.")
```

### 5. Evaluación de Veracidad implícita (*Truthy* y *Falsy*)
En Python, no siempre necesitas comparar explícitamente con `== True` o `== False`. Muchos objetos tienen un valor booleano implícito en un `if`.

Los siguientes elementos se evalúan siempre como **`False`**:
* El valor especial `None`.
* El booleano `False`.
* El cero de cualquier tipo numérico (`0`, `0.0`, `0j`).
* Secuencias y colecciones vacías (`""`, `[]`, `()`, `{}`, `set()`).

Cualquier otro objeto fuera de esta lista se evalúa de forma nativa como **`True`**.

```python
nombres = []  # Lista vacía

if nombres:
    print("Hay usuarios registrados.")
else:
    print("La lista está vacía.")  # Se ejecuta este bloque
```

La sentencia `match` evalúa una expresión y la compara secuencialmente con diferentes patrones definidos en bloques `case`. En cuanto coincide con uno, ejecuta su código y finaliza la evaluación.

### 1. Sintaxis Básica (Equivalente a Switch)
Para comparaciones de valores literales simples. El carácter guion bajo `_` actúa como el caso comodín por defecto (*wildcard*), equivalente al `default`:

```python
estado_usuario = "inactivo"

match estado_usuario:
    case "activo":
        print("El usuario está en línea.")
    case "inactivo":
        print("El usuario no está disponible.")
    case _:
        print("Estado desconocido.")
```

### 2. Coincidencia Múltiple (Operador OR `|`)
Puedes agrupar varios valores en un solo bloque `case` utilizando la barra vertical `|`:

```python
dia = "Sábado"

match dia:
    case "Lunes" | "Martes" | "Miércoles" | "Jueves" | "Viernes":
        print("Es un día laboral.")
    case "Sábado" | "Domingo":
        print("Es fin de semana.")
```

### 3. Desempaquetado y Captura de Patrones
Permite validar la estructura de colecciones (como listas o tuplas) y asignar variables internas dinámicamente:

```python
comando = ["mover", "izquierda", 10]

match comando:
    # Coincide si la lista tiene un solo elemento
    case ["salir"]:
        print("Cerrando aplicación...")
    # Coincide si tiene 3 elementos y el primero es "mover"
    # Captura los otros dos elementos en las variables 'direccion' y 'pasos'
    case ["mover", direccion, pasos]:
        print(f"Moviendo {pasos} unidades hacia la {direccion}.")
    case _:
        print("Comando no reconocido.")
```

### 4. Condicionales Internos (Guards con `if`)
Puedes añadir una cláusula `if` a un `case` para aplicar un filtro lógico adicional antes de dar el patrón por válido:

```python
pedido = {"producto": "laptop", "cantidad": 5}

match pedido:
    case {"producto": nombre, "cantidad": cant} if cant > 3:
        print(f"Pedido mayorista de {cant} unidades de {nombre} (Aplica descuento).")
    case {"producto": nombre, "cantidad": cant}:
        print(f"Pedido estándar de {cant} unidades de {nombre}.")
```

### 5. Validación por Tipo de Dato
Permite comprobar si un objeto pertenece a una clase específica (como `int`, `str`, etc.) e incluso capturar su valor:

```python
dato = 42

match dato:
    case int(valor):
        print(f"Es un número entero: {valor}")
    case str(valor):
        print(f"Es una cadena de texto: '{valor}'")
    case _:
        print("Es otro tipo de dato.")
```

## Bucles `while`

Al igual que con el `if`, el cuerpo de un bucle `while` se define mediante la **indentación**. Si la condición inicial es falsa desde el principio, el bloque de código nunca llegará a ejecutarse.

### 1. Estructura Básica y Sintaxis
El bucle evalúa la condición antes de cada iteración. Es fundamental asegurar que la condición cambie dentro del bucle para evitar bucles infinitos:

```python
contador = 1

while contador <= 3:
    print(f"Iteración número: {contador}")
    contador += 1  # Modifica la condición para terminar el bucle
```

### 2. Control de Bucles: `break` y `continue`
* **`break`**: Rompe el bucle de manera inmediata y transfiere el control a la línea de código posterior al `while`.
* **`continue`**: Omite el resto del código de la iteración actual y salta directamente a evaluar la condición del bucle nuevamente.

```python
# Ejemplo de break
while True:
    respuesta = input("Escribe 'salir' para terminar: ")
    if respuesta == "salir":
        break

# Ejemplo de continue
n = 0
while n < 5:
    n += 1
    if n == 3:
        continue  # Se salta el print para el número 3
    print(n)
```

### 3. La Cláusula Opcional `else` en Bucles
Python permite asociar un bloque `else` a un bucle `while`. Este bloque se ejecuta **únicamente cuando el bucle termina de forma normal** (porque la condición pasó a ser falsa). 

⚠️ **Importante:** Si el bucle se interrumpe de forma abrupta mediante un **`break`**, el bloque `else` se ignora por completo.

```python
intento = 1

while intento <= 3:
    print(f"Intento de conexión {intento}...")
    # Imaginemos que aquí la conexión falla
    intento += 1
else:
    print("No se pudo establecer la conexión tras 3 intentos.")
```

### 4. Bucles Infinitos Intencionados
Es común utilizar `while True:` combinándolo con un condicional `if` y un `break` interno. Es un patrón muy utilizado para menús de consola o procesos que deben escuchar eventos de forma indeterminada:

```python
while True:
    print("\n--- MENÚ ---")
    print("1. Saludar")
    print("2. Salir")
    opcion = input("Elige una opción: ")
    
    if opcion == "1":
        print("¡Hola!")
    elif opcion == "2":
        print("Adiós.")
        break
    else:
        print("Opción inválida.")
```

## Estructura de las funciones en python

Las funciones en Python se definen con la palabra clave `def`, seguida del nombre de la función, sus parámetros entre paréntesis `()` y dos puntos `:`. El cuerpo se delimita mediante la indentación.

### 1. Sintaxis Básica y Valores por Defecto
Puedes asignar valores predeterminados a los parámetros. Si el usuario no proporciona un argumento al llamar a la función, se utilizará el valor por defecto.

⚠️ **Regla de oro:** Los parámetros con valores por defecto **siempre deben ir al final** de la lista de parámetros.

```python
def saludar(nombre, saludo="Hola"):
    print(f"{saludo}, {nombre}!")

saludar("Carlos")           # Salida: Hola, Carlos!
saludar("Ana", "Buenos días") # Salida: Buenos días, Ana!
```

### 2. Número de Argumentos Variables
Python permite que una función reciba una cantidad indeterminada de argumentos mediante dos operadores:

* **Posicionales Variables (`*args`):** Captura los argumentos sobrantes como una **tupla**.
* **Nominales Variables (`**kwargs`):** Captura los argumentos con nombre sobrantes como un **diccionario**.

```python
def super_funcion(*args, **kwargs):
    print(args)   # Tupla de argumentos posicionales
    print(kwargs) # Diccionario de argumentos por nombre (llave=valor)

super_funcion(1, 2, 3, usuario="admin", activo=True)
# Salida args: (1, 2, 3)
# Salida kwargs: {'usuario': 'admin', 'activo': True}
```

### 3. ¿Paso por Valor o por Referencia?
En Python **no existe** técnicamente el paso por valor ni por referencia tradicional. Se utiliza un mecanismo llamado **"Paso por asignación de objetos"** (*Pass by object reference*). Su comportamiento depende de si el objeto es mutable o inmutable:

* **Objetos Inmutables (`int`, `str`, `tuple`, `float`, `bool`):** Se comportan similar a un **paso por valor**. Si modificas el parámetro dentro de la función, la variable externa original **no cambia**.
* **Objetos Mutables (`list`, `dict`, `set`):** Se comportan similar a un **paso por referencia**. Si modificas el contenido del objeto dentro de la función, los cambios **afectarán a la variable externa**.

```python
# Ejemplo con Inmutable (int)
def modificar_entero(n):
    n = 10

x = 5
modificar_entero(x)
print(x) # Salida: 5 (No cambia)

# Ejemplo con Mutable (list)
def modificar_lista(L):
    L.append(4)

mi_lista = [1, 2, 3]
modificar_lista(mi_lista)
print(mi_lista) # Salida: [1, 2, 3, 4] (¡Sí cambia!)
```

### 4. Valores Devueltos por una Función (`return`)
Las funciones utilizan la palabra clave `return` para enviar valores de vuelta al código que las invocó.

* **Retorno implícito:** Si una función no tiene una instrucción `return`, o tiene un `return` vacío, devuelve automáticamente **`None`**.
* **Retornos múltiples:** Python permite devolver varios valores separados por comas. Técnicamente, empaqueta los elementos y los retorna como una **tupla**, la cual se puede desempaquetar directamente.

```python
# Retorno múltiple
def calcular_estadisticas(a, b):
    suma = a + b
    multiplicacion = a * b
    return suma, multiplicacion  # Retorna la tupla (suma, multiplicacion)

resultado_suma, resultado_mult = calcular_estadisticas(5, 3)
print(resultado_suma) # Salida: 8
print(resultado_mult) # Salida: 15
```

````python
def quickSort(l):
    if l==[]:
        return l
    l1=[]
    l2=[]
    a=l[0]
    for x in l[1:]:
        if x<=a:
            l1.append(x)
        else:
            l2.append(x)
    return quickSort(l1)+[a]+quickSort(l2)
````

## Decoradores

````python
def miDecorador(funcion):
    #print('Llamada a miDecorador para construir una nueva función',funcion.__name__)
    def nueva(*args):
        print("Entrando en función",funcion.__name__," - argumentos:",args)
        ret=funcion(*args)
        print("Hemos acabado con la función",funcion.__name__)
        print("Valor devuelto por",funcion.__name__,"es",ret)
        return ret
    return nueva
    #return funcion
````

## Ficheros de texto

Tenemos tres formas de abrir un fichero de texto: 'r'
(modo lectura), 'w' (modo escritura), 'a' (modo añadir).
La orden para abrir un fichero es `open` y su
sintaxis es
> `file = open( ruta_hasta_el_fichero, modo_apertura)`

Los modos de apertura hacen lo siguiente:
* 'r': Abre el fichero para lectura y sitúa la posición de lectura al comienzo del mismo. El fichero ha de existir.
* 'w': Abre el fichero para escritura y sitúa la posición de escritura al comienzo del mismo. Si existe el fichero lo sobreescribe. Si el fichero no existe lo crea.
* 'a': Abre el fichero para escritura y sitúa la posición de escritura al final del mismo. Si el fichero no existe lo crea.

Para escribir en un fichero tenemos la orden `write`:
> `file.write('hola')`

Para la lectura tenemos `read()` que lee el fichero completo:
> `cadena=file.read()`

Por último, al finalizar de trabajar con un fichero debemos cerrarlo con la orden `close`:
> `file.close()`

Para almacenar variables en formato binario usamos la librería [struct](https://docs.python.org/3/library/struct.html#struct-format-strings) y las funciones `pack` y `unpack`

````python
import struct
````

````python
help(struct.pack)
help(struct.unpack)
````

````python
import struct, random
lfloat=[random.random() for x in range(5)]
print(lfloat)
lbytes=struct.pack( "%if" % len(lfloat), *lfloat )
print( type( lbytes), len(lbytes) )
struct.unpack( "%if" % len(lfloat), lbytes )
````

Indicamos con una cadena, por ejemplo "5d",  el formato y la cantidad de datos que queremos pasar a binario (en nuestor ejemplo cinco valores de tipo `double`)

````python
import struct, random
lfloat=[random.random() for x in range(5)]
print(lfloat)
lbytes=struct.pack( "{}d".format( len(lfloat) ), *lfloat )
print( type( lbytes), len(lbytes) )
struct.unpack( "{}d".format(len(lfloat)), lbytes )
````

````python
import requests
def obtenerDatosDeIndice(indice,anioI,mesI,anioF,mesF,acumularEnTrimestres=False):
    '''
    Obtiene una lista de valores para un índice dado entre dos fechas. 
    Si acumularEnTrimestres es True acumula los valores por trimestre
    anioI, mesI: año y mes inicio
    anioF, mesF: año y mes final
    '''
    if mesI<10:
        mesI='0'+str(mesI)
    if mesF<10:
        mesF='0'+str(mesF)
    direccion=crearURL('ES','DATOS_SERIE',indice,'date={}{}01:{}{}31'.format(anioI,mesI,anioF,mesF)) #'nult=3')
    print("dirección a la que me conecto:",direccion)
    cnx=requests.get(direccion)
    diccionario=cnx.json()
    cnx.close()
    # Vemas la diferencia en segundos entre las dos primeras fechas
    deltaFechas=(diccionario['Data'][1]['Fecha']-diccionario['Data'][0]['Fecha'])/1000
    daux=[x['Valor'] for x in diccionario['Data']]
    # sólo podemos acumular en trimestres si los datos son mesuales
    # deltaFechas <= 3600*24*31 ocurre si los datos son mensuales
    if acumularEnTrimestres and deltaFechas <= 3600*24*31:
        daux = [sum(daux[i:i+3]) for i in range(0,len(daux),3)]
    return daux
````

````python
from random import randint

def creaListaAleatoria(tam):
    lnumeros=[]
    for i in range(tam):
        x=randint(0,9999999999)*(-1)**(randint(0,1))
        lnumeros.append(x)
    return lnumeros
````

````python
def compruebaArgumentos(a1,a2):
    correcto=True
    tam=0

    try:
        tam=int(a1)
        if not(0<=tam and tam<=10000):
            correcto=False
            print('El número no está entre 0 y 10000')
            return
    except:
        correcto=False
        print('No es entero')
        return

    if correcto:
        aux=a2[-4:]
        if aux!='.txt':
            a2=a2+'.txt'

    return (tam,a2)
````
