```python
def pintaMenu(listaOpciones):
    '''
    Función para pintar un menú
    Argumentos:
        listaOpciones -> lista con opciones del menú
    Salida:
        número entre 0 y número de opciones del menú
    Ejemplo:
        > pintaMenu(["Introducir api_key","opción 2","opción 3"])
    '''
    while True:
        print()
        for i,opcion in enumerate(listaOpciones):
            print(f"{i+1}.-{opcion}")
        print(f"0.- Salir")
        cad=input("Opcion: ")
        try:
            k=int(cad)
            if 0<=k and k<=len(listaOpciones):
                return k
            else:
                print(f"El valor ha de estar entre 0 y {len(listaOpciones)}")
        except:
            print("Opción no válida")
```
